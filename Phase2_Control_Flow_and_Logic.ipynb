### **Module 2.1: Conditional Statements (if, elif, else).**

Every program we wrote ran every single line from top to bottom without thinking.

In real life, programs need to make decisions:

* *If* the user enters the right password, log them in; *otherwise (`else`)*, show `"Access Denied"`.
* *If* the cart total is above ₹500, make delivery ₹0; *otherwise*, add ₹40.

---

* If the condition is **`True`**, Python enters the `if` block.
* If the condition is **`False`**, Python skips the `if` block and goes into the `else` block.


In Python, we teach our code to make decisions using if, elif, and else.

### 1. The Simple `if - else` (Two Paths)

Think of `if - else` as a fork in the road. Python checks a condition:

* If the condition is **`True`**, Python enters the `if` block.
* If the condition is **`False`**, Python skips the `if` block and goes into the `else` block.


Look closely at the grammar (syntax) below:

````python
age = int(input("Enter your age: "))if age >= 18:

print("You are eligible to vote!")else:

print("Sorry, you are too young to vote.")
````

**Two Strict Grammar Rules You Must Notice Above:**

1. **The Colon (`:`):** Every `if`, `elif`, and `else` line **MUST end with a colon `:`**. Think of `:` as saying *"Then do the following:"*.
2. **Indentation (The 4-Space Gap):** Look at the `print()` lines—they are pushed slightly to the right (by 4 spaces or 1 `Tab` key press).
   * In Python, this gap is called **Indentation**.
   * Indentation tells Python **which lines belong inside the `if` or `else` block**.
   * When you press `Enter` after typing a colon `:` in Google Colab, Colab automatically adds that 4-space gap for you! If you remove that gap, Python throws an `IndentationError`.

### 2. Multiple Conditions: if - elif - else (More Than Two Paths)

What if there are 3 or more possibilities? (Like Traffic Lights: Red, Yellow, Green—or Grading: A, B, C, Fail).

In English, we say "else if". Python shortens "else if" into elif.

````python
marks = int(input("Enter your marks: "))

if marks >= 90:
  print("Grade: A")
elif marks >= 75:
  print("Grade: B")
elif marks >= 50:
  print("Grade: C")
else:
  print("Result: Fail")
````

**How Python executes this (Top-to-Bottom Chain):**



1. Suppose you enter `80`.
2. Python checks `if marks >= 90:` -> `80 >= 90` is `False`. It skips this block.
3. Python moves to the first `elif marks >= 75:` -> `80 >= 75` is **`True`**! It prints `Grade: B`.
4. **Crucial Rule:** The moment ONE branch in an `if-elif-else` chain becomes `True`, Python runs that block and **skips the entire rest of the chain**! Even though `80 >= 50` is also true, Python never even checks it because `elif marks >= 75` already won.
5. If **none** of the `if` or `elif` conditions are `True` (say, `marks = 30`), Python automatically runs the final `else:` catch-all block.

### 3. Nested if (if inside another if)

Sometimes, you only check a second condition after the first condition passes—like checking an ATM PIN only after verifying that the card is valid.

You just **indent one more level** inward:

````python
email = input("Enter email: ")

if "@" in email:
    password = input("Enter password: ")
    if password == "1234":
        print("Login Successful!")
    else:
        print("Wrong Password!")
else:
    print("Invalid Email Address!")
````

### Advanced Tricks and Concepts

#### 1. Truthiness and Falsiness (Implicit Booleans)

In Python, an `if` condition does not strictly need to be a `True` or `False` boolean expression. Python automatically judges certain values as **Falsy** (treated as `False`) or **Truthy** (treated as `True`):
* **Falsy values:** `0`, `0.0`, `""` (empty string), `[]` (empty list), `None`, and `False`.
* **Truthy values:** Everything else! (Any non-empty string, any non-zero number).

````python
name = input("Enter your name: ")

# Instead of writing: if name != "":
if name:
    print("Welcome,", name)
else:
    print("You forgot to type your name!")
````

#### 2. Short-hand if-else (The Ternary Operator)
If you have a simple if-else statement to assign a value, Python lets you write it on a single line:

````python
# Standard way:
if marks > 40:
    result = "Pass"
else:
    result = "Fail"

# Short-hand Ternary way:
result = "Pass" if marks > 40 else "Fail"
````

#### 3. The pass Keyword (The Placeholder)
If you leave an if block completely empty, Python will throw an IndentationError. To prevent this, use pass as a dummy placeholder that tells Python to skip it without crashing, it literally means "do nothing":

````python
if age >= 18:
    pass   # Keeps Python happy until you write the actual code later
````

### Module 2.2: The `while` Loop - Repeating an action until a condition changes.

A loop is used to repeat a block of code multiple times automatically. A `while` loop keeps executing its indented code block **as long as** its condition remains `True`. Once the condition becomes `False`, the loop stops.

**The 3 Pillars of Every `while` Loop:**
1. **Initialization (Start):** Setting a starting value for the loop variable before the loop begins.
2. **Condition (Stop):** The rule checked before every round.
3. **Update (Step):** Changing the variable inside the loop so it moves toward the stop condition.

**Example: Counting from 1 to 5**


````python
i = 1              # 1. Start
while i <= 5:      # 2. Stop Condition
    print(i)
    i = i+1         # 3. Step (Update)
````

**The Infinite Loop Trap:** <br>
If you forget to update the variable inside the while loop (e.g., forgetting i += 1), the condition remains True forever, and the loop never stops. In Google Colab, you must click the Stop (■) button if you accidentally trigger an infinite loop.

````python
num = int(input("Enter a number: "))

while num > 0:
    last_digit = num % 10    # Step 1: Extract the last digit
    print(last_digit)        # Step 2: Print it
    num = num // 10          # Step 3: Chop off the last digit (Update Step)
````

````python
jackpot = 7
guess = int(input("Guess the secret number (1-10): "))

while guess != jackpot:
    print("Wrong guess, try again!")
    # The Update Step: We must ask for a NEW guess inside the loop!
    guess = int(input("Guess again: "))

print("You won! The number was 7.")
````

### Module 2.3: The `for` Loop and `range()`

A `for` loop is used to iterate over a sequence of numbers (or items). In Python, we usually pair `for` loops with the `range()` function.

#### How `range()` works:
`range(start, stop, step)` generates numbers from `start` up to (but **not including**) `stop`.

* **`range(5)`** -> Generates `0, 1, 2, 3, 4` (Starts at `0` by default, stops before `5`).
* **`range(1, 6)`** -> Generates `1, 2, 3, 4, 5` (Starts at `1`, stops before `6`).
* **`range(1, 10, 2)`** -> Generates `1, 3, 5, 7, 9` (Starts at `1`, jumps by `2`, stops before `10`).

#### Example 1: Printing numbers 1 to 5


````python
for i in range(1, 6):
    print(i)
````

#### Example 2: Counting Backwards
To count backwards, your step must be negative, and you start at a high number and stop at a lower number: <br>
(This prints 5, 4, 3, 2, 1. It stops before hitting 0)

range(5,0,-1)<br>
three parts: **start, stop, and step.**

````python
for i in range(5,0,-1):
  print(i)
````

### Understanding `range(2, 11, 2)`

Let's break down `range(2, 11, 2)` into its three parts: `start`, `stop`, and `step`.

* **Start (`2`):** The loop begins at exactly `2`.
* **Stop (`11`):** The loop must stop *before* it hits `11`.
* **Step (`2`):** It jumps by `2` every round.

### How Python executes it, step by step

1. Starts at `2` → prints `2`
2. Jumps by 2 → `4` → prints `4`
3. Jumps by 2 → `6` → prints `6`
4. Jumps by 2 → `8` → prints `8`
5. Jumps by 2 → `10` → prints `10`
6. The next jump would be `12`, but `12` is past the stop limit of `11`, so the loop stops!

### Module 2.4: Loop Control (`break` and `continue`)

Sometimes you need to alter a loop's normal flow based on a specific condition.

Python gives you two emergency buttons:

* **`break`:** Completely shatters the loop. Python immediately exits the loop and moves on to the rest of the program.


````python
  for i in range(1, 10):
      if i == 5:
          break      # The loop completely dies when i is 5
      print(i)       # Output: 1, 2, 3, 4
````

* **`continue`:** Skips the current iteration. Python ignores the remaining code in the loop body for that round and jumps straight back to the top for the next round.


````python
for i in range(1, 6):
    if i == 3:
        continue   # Skips printing 3 and jumps to i = 4
    print(i)       # Output: 1, 2, 4, 5
````

### Summary: `break`, `continue`, and `pass`

* **`break`:** Kills the loop completely.
* **`continue`:** Skips the `print` statement for that one round, but keeps the loop alive.
* **`pass`:** Does absolutely nothing; the code acts like the `if` statement isn't even there.
