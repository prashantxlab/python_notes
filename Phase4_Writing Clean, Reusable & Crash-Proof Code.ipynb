# Phase 4: Writing Clean, Reusable & Crash-Proof Code

Up until now, we have been writing scripts that run once from top to bottom. But what if you write a brilliant piece of code—like a 5-line loop that cleans up phone numbers—and you need to use it 50 times across your program? Copying and pasting those 5 lines everywhere makes your code massive, messy, and hard to fix if you find a bug.

Functions solve this. They allow you to package a block of code into a custom, reusable tool. You have actually been using built-in functions since Day 1: `print()`, `len()`, and `type()`. Now, it is time to build your own.

### Module 4.1: Functions (def) — Packaging code for reuse

#### 1. Defining and Calling Functions

A **Function** is a reusable block of code that only runs when it is explicitly called.
* **Define (`def`):** The blueprint. Use the `def` keyword, followed by a name, parentheses `()`, and a colon `:`. <br>
**Note:** Defining a function does not actually run the code inside it.
* **Indentation:** Everything inside the function must be indented (just like `if` statements and `for` loops).
* **Call:** To execute the function, type its name followed by parentheses `()`.

```python
# 1. Defining the function (Python memorizes this, but doesn't run it yet)

def greet_user():
  print("Welcome to the system!!!")
  print("Your login was successful.!!!")

print("----System Starting----")

# 2. Calling the function (This actually makes the code run)
greet_user()

print("---- Doing Other things ----")

# You can call it as many times as you want!
greet_user()
```

#### 2. Parameters and Arguments
Our greet_user() function is a bit dumb—it says the exact same thing every time. Real machines take Inputs (raw materials) and give you a specific Output.

To make a function flexible, you can pass data into it.
* **Parameter:** An empty variable placed inside the parentheses when you define the function. It acts as a placeholder.
* **Argument:** The actual data passed into the parentheses when *calling* the function.

You can add as many parameters as you want, separated by commas.

```python
# 'name' is the Parameter (the placeholder)
def greet(name):
  print(f"Hello {name}, welcome back!!!")

greet("Prashant")
greet("Kumar")
```

```python
# Function with MULTIPLE parameters

def calculate_total(price,tax):
  total = price + tax
  print(f"the final price is: ${total}")

calculate_total(100, 18)
calculate_total(500, 50)
```

#### 3. The Big Trap: return vs print

When you use `print()` inside a function, Python displays the result on the screen for your eyes, but the program's memory instantly forgets it. You cannot save that result into a variable or use it for later math.<br>
Analogy: print is like a restaurant showing you a picture of a pizza. return is the kitchen actually handing you the pizza in a box to take home.
* **`print()`:** Displays data to the human user. The program does not remember the value. If you try to save a printed function to a variable, it stores `None`.
* **`return`:** Gives the data back to the program. You can save the result in a variable, use it in an `if` statement, or do more math with it.
* **Exit Point:** A `return` statement immediately stops the function. No code below it will execute.



```python
# Function 1: Uses print (The program forgets the result)
def add_with_print(a,b):
  total = a + b
  print("Printed Total",total)

# Function 2: Uses return (The program remembers the result)

def add_with_return(a,b):
  total = a + b
  return total

print("--- Testing Print ---")
result1 = add_with_print(5,5)
print("what did the variable save?", result1)

print("\n---- Testing Return ----")
result2 = add_with_return(5,6)
print("what did the variable save?",result2)

# Because result2 actually holds '10', we can do more math with it!
print("Result2 multiplied by 2:", result2*2)
```

### Module 4.2: Variable Scope & `*args` / `**kwargs`

#### 1. Variable Scope (Local vs Global)

**Scope** means *where* a variable can be seen and used.

Think of a function as a **hotel room** and the main program as the **hotel lobby**:

* **Local scope (the room):** A variable created inside a function exists only in that function. When the function finishes, the room is cleaned out and the variable is destroyed. Nobody outside can see it.
* **Global scope (the lobby):** A variable created outside any function is visible everywhere, including inside functions. They can **read** it.

| | Local Variable | Global Variable |
|---|---|---|
| Created | Inside a function | Outside all functions |
| Visible to | Only that function | The whole program |
| Lifetime | Until the function ends | Until the program ends |
| Used outside its area | `NameError` | Works fine |

**Rule of thumb:** a function can look *out* into the lobby, but the lobby can never look *into* a function's room.

```python
# GLOBAL variable: lives outside, so every function can read it
discount_rate = 0.10

def calculate_price(price):
  final = price-(price * discount_rate)    # LOCAL variable: exists only while this function runs
  return final

print("Final Price:", calculate_price(100))
```

#### 2. Flexible Function Inputs: `*args` and `**kwargs`

Normally a function expects an exact number of inputs. But what if you don't know in advance how many values will be passed? Python solves this with two special tools.

* **`*args`:** Allows a function to accept an unlimited number of standard arguments. Python packs them into a **Tuple**.
* **`**kwargs`:** Allows a function to accept an unlimited number of named (keyword) arguments. Python packs them into a **Dictionary**.

| | `*args` | `**kwargs` |
|---|---|---|
| Stands for | Arguments | Keyword arguments |
| You pass | `10, 20, 30` | `name="Asha", age=20` |
| Python stores it as | Tuple | Dictionary |
| Loop with | `for x in args` | `for k, v in kwargs.items()` |

**Note:** The names `args` and `kwargs` are just a convention. The real magic is the `*` and `**`.

```python
# *args: any number of values, stored in a tuple
def add_all(*args):
  print("args is:", args)
  return sum(args)

print(add_all(10,20))
print(add_all(1,2,3,4,5))

# **kwargs: any number of name=value pairs, stored in a dictionary
def show_profile(**kwargs):
  print("kwargs is:",kwargs)
  for key, value in kwargs.items():
    print(key,"->",value)

show_profile(name="Asha",age=20, city="Delhi")
```

```python
# 1. *args (Packs unlimited inputs into a Tuple)
def  calculate_bill(*prices):
  print(f"Python packed the prices into a: {type(prices)}")
  return sum(prices)

print("Total for 2 items:", calculate_bill(100, 50))
print("Total for 4 items:", calculate_bill(100,50,200,10))

print("\n-------------------------------\n")

# 2. **kwargs (Packs named inputs into a Dictionary)
def build_profile(**info):
  print("Profile data stored as:", type(info))
  for key, value in info.items():
    print(f"{key} -> {value}")

# You can pass as many custom labels as you want

build_profile(name="Prashant",company="Quantix Innovations",location="Bengaluru")
```

### Module 4.3: Lambda, map(), and filter()

Writing compact one-line functions.

#### 1. Lambda (The One-Line Function)

A **lambda** is a small, nameless function written in a single line. Use it for quick, simple jobs that you don't want to write a full `def` block for.

**Syntax:** `lambda inputs: output`

#### 1. Lambda Functions (Anonymous Functions)
A `lambda` function is a small, unnamed function written in a single line. It is used for quick, throwaway operations where creating a full `def` block is overkill.
* **Syntax:** `lambda arguments: expression`

#### 2. `map()` and `filter()`
These are built-in tools used to process collections of data instantly. They are almost always paired with `lambda` functions to avoid writing standard `for` loops.
* **`map(function, iterable)`:** Runs the function on *every single item* in the list and returns the modified items.
* **`filter(function, iterable)`:** Runs a condition test on every item and returns *only* the items that pass (`True`).

*(Note: Both `map` and `filter` output raw memory objects, so you must wrap them in `list()` to see the actual output array).*

* **`lambda` (The Sticky Note):** A quick, 1-line function without a formal `def` name.
  * Formula: `lambda input: what_to_do_with_input`
* **`map()` (The Bartender):** Takes a list and does the **exact same math/action** to every single item. (Example: Add 10 bonus marks to every student).
* **`filter()` (The Bouncer):** Takes a list and applies a `True/False` test. It **removes** items that fail the test and keeps the ones that pass. (Example: Keep only students who scored above 40).

```python
# 1. Lambda Function (A 1-line function to double a number)
double = lambda x: x*2
print("lambda output:", double(50))
```

```python
prices = [100, 200, 300, 400, 500]

# 2. map() - Apply a 5% tax to every price in the list
# The lambda takes 'p' and returns 'p * 1.05'
taxed_prices = list(map(lambda p: p*1.05, prices))
print("Mapped Prices:", taxed_prices)

# 3. filter() - Keep only the prices strictly greater than 250
# The lambda takes 'p' and returns True if p > 250, False otherwise
high_prices = list(filter(lambda p: p>250, prices))
print("Filtered Prices:", high_prices)
```

```python
# Our raw data
marks = [35,80,42,28,95]

# 1. map() -> The Bartender
# Give every single student 5 extra grace marks.
# The sticky note (lambda) says: take 'm' and return 'm + 5'
new_marks = list(map(lambda m: m+5, marks))
print("Marks after map():", new_marks)

# 2. filter() -> The Bouncer
# Only keep the students who scored 40 or higher.
# The sticky note (lambda) says: test if 'm >= 40'
passed_students = list(filter(lambda m: m>=40, marks))
print("Marks after filter():", passed_students)

passed_students_grace = list(filter(lambda m: m>=40, new_marks))
print("Marks after filter():", passed_students_grace)
```

### Module 4.4: Exception Handling (`try`, `except`, `finally`)

Right now, if your code asks a user to input their age, but they accidentally type the word "Twenty" instead of the number 20, Python panics. It throws a giant red error `(ValueError)` and instantly crashes your entire program.

**Exception handling** prevents your program from crashing when unexpected errors occur.

* **`try`:** The block where you place the code that might fail or crash.
* **`except`:** The safety net. If an error occurs in the `try` block, the program jumps here instead of stopping.
* **`finally`:** A block that executes no matter what, regardless of whether an error happened or not.

```python
# A simple division program that could crash if the user enters 0

denominator = 0

print("---Program Started---")

try:
  # Risky code: Dividing by zero causes a ZeroDivisionError
  result = 100/denominator
  print(f"The result is {result}")

except ZeroDivisionError:
  # Python jumps here instead of crashing
  print("Error: You can not divide a number by zero!!!")

except Exception as e:
  # This catches any other random errors we didn't plan for
  print(f"An unknown error occured: {e}")

finally:
  # This runs no matter what
  print("Cleanup: Closing the math operation.")

print("--- Program is still running! ---")
```
