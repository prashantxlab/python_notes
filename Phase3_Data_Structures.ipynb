### Module 3.1: Strings In-Depth

In Phase 1, we learned that anything inside quotes ("Hello") is a String (str).
In real-world programs, user data comes in as messy text—extra spaces in emails, full names that need to be split into first and last names, or phone numbers with +91- that need to be cleaned up.

To manipulate text, Python treats a String as an ordered sequence of individual characters placed in numbered boxes.

### 1. Indexing: Picking a Single Character <br>
Every single character in a string—including letters, numbers, symbols, and even blank spaces—gets a position number called an Index.

Python gives every character **two index numbers**:

* **Positive Indexing (Left to Right):** Always starts at `0` for the first character.
* **Negative Indexing (Right to Left):** Always starts at `-1` for the very last character. (Super useful when you want the last letter of a word without counting how long the word is!)

### How `"PYTHON"` is stored in memory

| Character | P | Y | T | H | O | N |
|---|---|---|---|---|---|---|
| **Positive Index** (Left $\rightarrow$ Right) | 0 | 1 | 2 | 3 | 4 | 5 |
| **Negative Index** (Right $\leftarrow$ Left) | -6 | -5 | -4 | -3 | -2 | -1 |

### Accessing a character

To grab a character, write the variable name followed by the index inside square brackets `[]`:

```python
word = "PYTHON"

print(word[0])   # P
print(word[3])   # H
print(word[-1])  # N (last character)
```

### 2. Slicing: Cutting a Piece of Text ([start:stop:step])

What if you don't want just one letter, but a whole chunk of text—like pulling "Camp" out of "CampusX"? This is called Slicing.

Remember range(start, stop, step) from for loops? Slicing uses the exact same logic, except with colons : inside square brackets:
variable `[start : stop : step]`

* **`start`:** Starting index (**Included**). If left blank, starts from the beginning (`0`).
* **`stop`:** Stopping index (**NOT Included** — stops 1 position before this). If left blank, goes all the way to the end.
* **`step`:** Jump size (Default is `1`). Using `-1` reverses the string.

### Shortcut rules for blank slots

* If you leave `start` blank (`word[:4]`), Python automatically starts from the very beginning (Index `0`).
* If you leave `stop` blank (`word[2:]`), Python automatically goes all the way to the very end of the string.
* **Famous interview trick** (`word[::-1]`): A step of `-1` reads the string backwards from right to left, reversing the entire word!




````python
word = "PYTHON"
#       012345
#     -6-5-4-3-2-1

print(word[:4])       #(Blank start means start from index 0)
print(word[2:])       #(Blank stop means go till the very end)
print(word[1:5])      #(start from 1 and stop before 5)
print(word[::2])      #(take the whole string, but jump 2 characters at a time)
print(word[::-1])     #(Step of -1 reverses the string)
````

### 3. String Formatting (f-strings)
### f-Strings (Formatted Strings)

In Phase 1 and Phase 2, whenever we wanted to print variables inside a sentence, we wrote:

```python
print("Hello", name, "you are", age, "years old.")
```

Notice how annoying it is to open and close quotes and type commas everywhere? Modern Python solves this with **f-strings** (Formatted Strings):

1. Type the letter `f` right before your opening quotation mark: `f"..."`
2. Inside the quotes, write your sentence normally, and whenever you want to plug in a variable (or even math!), just wrap it inside curly brackets `{}`:

```python

```

Python automatically swaps `{name}` and `{age}` with their actual values!

### Example


````python
name = "Asha"
age = 20

print(f"Hello {name}, you are {age} years old.")
print(f"Next year you will be {age + 1}.")
````

### 4. Common String Methods (`.strip()`, `.replace()`, `.split()`)

Python comes with built-in tools called **Methods** designed specifically to clean and transform text. You trigger a method by typing a dot `.` after your string variable: `variable.method()`.

* **Crucial Rule (Strings are Immutable):** In Python, strings cannot be changed in-place. When you use `.replace()` or `.strip()`, Python does not alter your original variable. It creates and outputs a **brand-new modified copy** of the string.

Here are the 3 most important methods in the syllabus:

1. **`.strip()` — The Space Cleaner:** Removes accidental blank spaces from the start and end of a string (for example, if a user types `" prashant@gmail.com "` into a login box). It does **not** remove spaces in the middle of a sentence.
2. **`.replace("old", "new")` — Find and Replace:** Searches for every occurrence of `"old"` in the string and swaps it with `"new"`.
3. **`.split("separator")` — The Chopper:** Cuts a string into a **list** of smaller words wherever it sees the separator.
   * If you leave the brackets empty (`.split()`), it chops at every blank space.
   * If you pass a comma (`.split(",")`), it chops at every comma.

### Example

```python
email = "  prashant@gmail.com  "
print(email.strip())                    # prashant@gmail.com

msg = "I like Java"
print(msg.replace("Java", "Python"))    # I like Python

fruits = "apple,banana,mango"
print(fruits.split(","))                # ['apple', 'banana', 'mango']
print("Learn Python fast".split())      # ['Learn', 'Python', 'fast']
```

### Proof that strings are immutable

```python
name = "  hello  "
name.strip()          # creates a new string, but we didn't save it
print(f"[{name}]")    # [  hello  ]  -> original is unchanged

name = name.strip()   # save the new copy back into the variable
print(f"[{name}]")    # [hello]
```

### Module 3.2: Introduction to Lists
Up until now, one variable could only hold one piece of data (mark = 85). If a class has 50 students, creating 50 separate variables (mark1, mark2, mark3...) is impossible to manage.

A List solves this by letting you store multiple items inside a single variable.
A List is a built-in data structure that allows you to store multiple items in a single variable.
* **Creation:** Created using square brackets `[]`.
* **Heterogeneous:** A single list can hold different data types (integers, strings, floats, booleans) at the exact same time.
* **Ordered (Indexing):** Every item in a list has a specific numbered position, always starting at `0`.
* **Mutable:** You can change, replace, or overwrite items after the list is created.

#### 1. Creating and Accessing Lists (Indexing)

````python
# A list holding different types of data
student = ["Rahul",24,8.5,True]
print(student)

# Positive Indexing (Starts at 0)
fruits = ["Apple","Banana","Cherry","Dates"]
print(fruits[0])  # First item -> Apple
print(fruits[2])  # Third item -> Cherry

# Negative Indexing (Starts at -1 from the end)
print(fruits[-1]) # Last item -> Dates
print(fruits[-2]) # Second-last item -> Cherry
````

### 2. Slicing Lists (Extracting a chunk)
Indexing (fruits[0]) pulls out one single item.
What if you want to pull out multiple items at once—like the middle two items? That is called Slicing.<br>
Slicing extracts a smaller list from a larger list. It works exactly like range(start, stop): it pulls items from the start index up to, but not including, the stop index. The grammar is `list_name[start:stop]`.

* **`start`:** Starting index (**Included**). If left blank (`[:3]`), starts from index `0`.
* **`stop`:** Stopping index (**NOT Included** — stops 1 position before this). If left blank (`[2:]`), goes until the very end.

````python
marks = [10, 20, 30, 40, 50, 60]

print(marks[1:4])   # Output: [20, 30, 40] (Indexes 1, 2, and 3)(stops before 4)
print(marks[:3])    # Output: [10, 20, 30] (If start is blank, it starts from the beginning)
print(marks[3:])    # Output: [40, 50, 60] (If stop is blank, it goes all the way to the end)
````

### 3. Mutability (Changing Data)
"Mutable" is just a fancy programming word that means "Changeable".<br>
Unlike basic variables where you have to overwrite the whole thing, you can target one specific slot in a list and change it.


*  In Module 3.1, we learned **Strings are Immutable (Locked)**. If you have word = "Bat", and you try to change the first letter "B" to "C" by writing word[0] = "C", Python crashes with an error. You cannot edit a single slot of a string.
*   **Lists are Mutable (Unlocked)**. You can open any box (index) inside a list, throw out the old item, and put a new item inside without creating a new list!

## Changing Items in a List

Look at how easy it is to change an item in a list:

1. Suppose you have: `colors = ["Red", "Blue", "Green"]`
2. `"Blue"` is sitting at Index `1`.
3. You want to replace `"Blue"` with `"Yellow"`.
4. You simply write: `colors[1] = "Yellow"`
5. Now `colors` becomes `["Red", "Yellow", "Green"]`!

````python
colors = ["Red", "Blue", "Green"]
print("Before change:", colors)

# Replace index 1 ("Blue") with "Yellow"
colors[1] = "Yellow"
print("After change:", colors)
````

### 4. Adding New Items to a List (.append() and .insert())

In Part 1, you used marks[0] = 95 to replace an item that was already in the list.
What if you want to add a brand-new item to make the list longer?

Python gives lists built-in commands called Methods. You use a method by typing a dot (.) right after your list's name: list_name.method()

Python gives Lists built-in tools (called Methods) to easily add, remove, and sort data.

### Methods in Python

A **method** is a built-in function that belongs to an object (like a string or list). You call it by typing a dot `.` after the object: `object.method()`.

```python
nums = [1, 2]
nums.append(3)
print(nums)           # [1, 2, 3]
```
####  Adding Items to a List (`.append()` and `.insert()`)

* **`.append(item)`:** Adds a single item to the **very end** of the list.
* **`.insert(index, item)`:** Inserts an item at a **specific index position** and shifts the remaining items to the right.

````python
cart = ["Milk", "Bread"]
print("Start", cart)

# 1. Add "Eggs" to the very end
cart.append("Eggs")
print("After append",cart)

# 2. Insert Butter at index 1
cart.insert(1,"Butter")
print("After insert at 1",cart)
````

### 5. Removing Items from a List (`.remove()` and `.pop()`)

What if a user deletes something from their shopping cart? Python gives you two ways to remove an item:

1. **`.remove(value)` — Delete by Value (Name):** Use this when you know what the item is, not its index number. Python searches the list from left to right and deletes the **first exact match**.
   * `cart.remove("Bread")` finds `"Bread"` and deletes it.
2. **`.pop(index)` — Delete by Index Number:** Use this when you want to delete an item sitting at a specific index position.
   * `cart.pop(0)` deletes whatever item is sitting at Index `0`.
   * **Empty Bracket Shortcut (`.pop()`):** If you leave the brackets completely empty (`cart.pop()`), Python automatically deletes the **very last item** in the list!

````python
items = ["Pen","Notebook","Eraser","Pencil"]
print("Start",items)

# 1. Remove by exact name/value
items.remove("Notebook")
print("After removing 'notebook':",items)

# 2. Remove by index position (Index 0 is 'Pen')
items.pop(0)
print("After pop(0):",items)

# 3. Remove the very last item using empty pop()
items.pop()
print("After empty pop()",items)
````

### 6. Counting and Sorting Lists (len(), .sort(), .reverse())

* **`len(list_name)` — Count Total Items:** Tells you how many items are inside the list using normal counting (`1, 2, 3...`). (Note: `len()` is a general Python function like `print()`, so the list name goes **inside the brackets** instead of using a dot.)
* **`.sort()` — Arrange in Order:** Rearranges the list from smallest to largest (for numbers) or A to Z (for words).
* **`.reverse()` — Flip Backwards:** Flips the current order of the list completely backwards.


````python
scores = [85, 40, 95, 60]

# 1. Count total items
print("Total subjects:", len(scores))

# 2. Sort from smallest to largest
scores.sort()
print("Sorted scores:", scores)

# 3. Flip the list backwards
scores.reverse()
print("Reversed scores:",scores)
````

### 7. Looping Through a List

When you have a list like fruits = ["Apple", "Banana", "Mango"] and want to print each item one by one, you do not need range(). You can tell a for loop to walk directly through your list:
```
fruits = ["Apple", "Banana", "Mango"]
for item in fruits:
    print(item)
```
#### How Python Runs This in Slow Motion

* **Round 1:** Python grabs the first item from `fruits` (`"Apple"`), puts it inside the variable `item`, and prints `Apple`.
* **Round 2:** Python grabs the second item (`"Banana"`), puts it inside `item`, and prints `Banana`.
* **Round 3:** Python grabs the last item (`"Mango"`), puts it inside `item`, and prints `Mango`.
* **No more items left**, so the loop stops automatically!

````python
fruits = ["Apple","Banana","Mango"]

for item in fruits:
  print(item)

# we can also do math on each item inside a loop:

prices = [100,200,300]
for p in prices:
  print("Prices with $10 tax",p+10)
````

### 8. List Comprehensions (The 1-Line Shortcut)

In real projects, you will often want to take an old list `(like nums = [1, 2, 3, 4])`, do some math on every number **(like multiplying by 2)**, and save the results into a `brand-new list ([2, 4, 6, 8])`.

Let's compare the Normal 4-Line Way with the 1-Line Shortcut **(List Comprehension)**:

````python
# This is four line code for above question

nums = [1,2,3,4]
doubled = []          # Step 1: Create an empty list
for n in nums:        # Step 2: Loop through each number
  doubled.append(n*2) # Step 3: Multiply by 2 and append to the new list
print(doubled)
````

````python
# The 1-Line Shortcut (List Comprehension):
# Python lets you pack Steps 1, 2, and 3 inside a single pair of square brackets []:

nums = [1,2,3,4]
doubled = [n*2 for n in nums]
print(doubled)
````

* **Formula:** `new_list = [action for item in old_list]`
  * **Left side (`action`):** What you want to do to each item (e.g., `n * 2`).
  * **Right side (`for item in old_list`):** The standard `for` loop header.

### Module 3.3: Tuples & Sets

A **Tuple** is exactly like a List, but with one massive difference: **Tuples are Immutable (Read-Only).**

Once you create a Tuple, its data is permanently locked. You cannot add new items, remove items, or overwrite existing slots. If you try, Python will immediately crash with a `TypeError`.

**Why use a Tuple instead of a List?** <br>If you are writing a program that stores fixed data, like the 7 days of the week, the latitude/longitude coordinates of Bengaluru, or standard tax rates, you put them in a Tuple. This protects the data so another programmer (or a bug in your code) cannot accidentally change `Monday` to `Funday`.

**Note:** Tuples use round brackets `()`, while lists use square brackets `[]`.

### 1. Tuples (Read-Only Lists)
A Tuple is a built-in data structure identical to a List, except it is **immutable** (cannot be changed after creation).
* **Creation:** Created using parentheses `()` instead of square brackets `[]`.
* **Use Case:** Used for constant data that must be protected from accidental modification.
* **Access:** Uses the exact same 0-indexed system and `[start:stop]` slicing as Lists.
* **No Modifying Methods:** Because they are locked, Tuples do not have `.append()`, `.insert()`, `.remove()`, or `.pop()`.

````python
# 1. Creating and Accessing a Tuple

days = ("Monday","Tuesday","Wednesday","Thursday")

print(days[0])
print(days[1:3])

# 2. The Immutability Rule (What you CANNOT do)

coordinates = (12.9716, 77.5946)
print(coordinates)

# 3. Utitliy functions still work
print("Total days:", len(days))
````

### 2. Sets (The Duplicate Destroyers)

A **Set** is a collection that uses curly braces `{}`. It has two massive differences from both Lists and Tuples:

1. **No Duplicates Allowed:** A Set automatically deletes any repeated items. If you try to store `[1, 1, 1, 2]`, the Set instantly crushes it down to `{1, 2}`.
2. **Unordered (No Indexing):** Sets do not have a guaranteed order. There is no Index `0` or Index `1`. If you try to run `my_set[0]`, Python will crash.

3. **Mutable (Sort of):** You can add or remove items, but you cannot change an existing item (because there are no index slots to target).

**Why use a Set?** <br> If a company has a list of 10,000 email addresses and wants to remove all duplicate sign-ups, doing it manually with a `for` loop takes a lot of code. With a Set, you just convert the list into a set, and the duplicates vanish instantly!


**Note:** An empty set must be written `set()`, because `{}` creates an empty dictionary.

**The Ultimate Trick:** You can convert a List into a Set to instantly destroy all duplicates, then convert it back into a List!

````python
emails = ["a@gmail.com", "b@gmail.com", "a@gmail.com", "c@gmail.com", "b@gmail.com"]

unique_emails = set(emails)

print(unique_emails)
print(len(unique_emails))
````

````python
# 1. Sets automatically remove duplicates

ids = {101,102,101,103,103}
print("Unique IDs:",ids)

# 2. Adding and Removing items

ids.add(104)
ids.remove(101)
print("After updates:", ids)

# 3. The Duplicate Destroyer Trick
emails = ["a@x.com","b@x.com","a@x.com"]

unique_emails = list(set(emails))
print("Cleaned List", unique_emails)
````

### Module 3.4: Dictionaries (Key-Value Pairs)

Lists are great for ordered items (`["Apple", "Banana"]`). But what if you want to store a complete profile of a user? If you use a list:

```python
user = ["Prashant", 25, "Bengaluru"]
```

You have to memorize that Index `0` is the name, Index `1` is the age, and Index `2` is the city. If the list has 50 items, it becomes a nightmare.

A **Dictionary** solves this by using **Key-Value Pairs**. Instead of using index numbers (`0, 1, 2`), you give every piece of data a custom label (a **Key**). Think of a real-life phonebook: you look up a Name (the Key) to find the Phone Number (the Value).

* **Syntax:** Uses curly braces `{}`, just like Sets, but every item must have a colon `:` separating the Key and the Value. Example: `{"name": "Prashant", "age": 25}`
* **Unordered & Indexed by Key:** Dictionaries do not use `0, 1, 2` indexing. You access a Value by calling its exact Key.
* **Mutable:** You can add new pairs, change existing values, or delete pairs.
* **Keys Must Be Unique:** You cannot have two `"age"` keys in the same dictionary. The second one will overwrite the first.

````python
student = {
    "name": "Prashant",
    "age": "25",
    "city": "Bengaluru"
}

# 2. Accessing a Value (Use the Key instead of an index number)
print("Student Name:",student["name"])
print("Student Age:",student["age"])

# 3. Updating an existing Value
student["city"]="Hyderabad"
print("Updated City:",student["city"])

# 4. Adding a brand-new Key-Value pair
student["grade"] = "A+"
print("Full Dictonary:", student)
````

### Dictionary Methods and Looping.
## 1. Extracting Keys, Values, and Items

Sometimes you don't want just one piece of data. You want to see all the labels or all the data at once. Python provides three essential methods:

* **`.keys()`:** Extracts all the keys.
* **`.values()`:** Extracts all the values.
* **`.items()`:** Extracts all key-value pairs.
* **`.pop(key)`:** Deletes a specific key-value pair.

**Looping:** To loop through a dictionary, use `.items()` to assign the key and value to two separate variables in the `for` loop.

````python
student = {"name": "Asha", "age":20, "city":"Delhi"}

print(student.keys())
print(student.values())
print(student.items())
````

## 2. Removing Key-Value Pairs

If a user deletes their account, you need to remove their data.

* **`.pop(key)`:** Removes the specific key and its value from the dictionary.

````python
student = {
    "name":"Asha",
    "age":20,
    "city":"Delhi"
}

print(student.keys())
print(student.values())

student.pop("city")
print(student)
````

## 3. Looping Through a Dictionary

Because a dictionary has two parts (Keys and Values), looping is slightly different than a list. You use `.items()` to grab **both the key and the value at the same time**: `for key, value in student.items():`


````python
student = {
    "name":"Asha",
    "age":20,
    "city":"Delhi"
}

#print(student)
#print(student["age"])

for key, value in student.items():
  print(key,"->",value)
````
