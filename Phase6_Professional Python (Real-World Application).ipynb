### Module 6.1: File Handling (Text Files)
File handling allows your program to save data permanently. We use the `with open()` statement because it automatically closes the file when the block finishes, preventing memory leaks and file corruption.

**The 3 Main Modes:**
* **`'r'` (Read):** Opens the file to read its contents. (Fails if the file does not exist).
* **`'w'` (Write):** Opens the file to write. Warning: This instantly erases everything in the file and starts fresh. If the file does not exist, it creates it.
* **`'a'` (Append):** Adds new data to the very end of the file without deleting the old data.

#### Reading Methods

* `f.read()`: the whole file as one string
* `f.readline()`: one line
* `for line in f:`: loop line by line (best for big files)

```python
# 1. WRITING to a file (Mode 'w')
# This creates 'server_logs.txt' in your Colab environment
with open("server_logs.txt","w") as file:
  file.write("System Boot Sequence Initiated. \n")
  file.write("Loading IAM Modules.... \n")

print("File created and written sucessfully!!!\n")

# 2. APPENDING to a file (Mode 'a')
# This adds to the file WITHOUT erasing the first two lines

with open("server_logs.txt","a") as file:
  file.write("Saviynt EIC connection established. \n")

# 3. READING from a file (Mode 'r')
with open("server_logs.txt", "r") as file:
  content = file.read()

print("--- Reading File Contents ---")
print(content)
```

#### The Three File Types

* **`.txt`:** Plain text. Good for notes and logs.
* **`.csv`:** Rows and columns, like Excel. Use the built-in `csv` module.
* **`.json`:** Stores dictionaries and lists exactly as Python sees them. Use the built-in `json` module. This is the most common format for apps and APIs.

| Task | JSON function |
|---|---|
| Python → file | `json.dump(data, f)` |
| File → Python | `json.load(f)` |

#### CSV and JSON Files

*   **CSV (Comma-Separated Values):** The standard for spreadsheet data. If you export a list of user accounts from a database, it usually comes out as a CSV. Python has a built-in csv module to read and write these row by row.

*   **JSON (JavaScript Object Notation):** The absolute standard for web APIs, REST calls, and configuration files. It looks exactly like a Python Dictionary. Python's built-in `json` module makes it incredibly easy to translate a dictionary into a `.json` file `(json.dump)` and read a `.json` file back into a dictionary `(json.load)`.

To work with tabular data or API data, Python provides built-in modules that do the heavy lifting for you.
* **`import csv`:** Used for reading/writing spreadsheet data row by row.
* **`import json`:** Used for API payloads and configs.
  * `json.dump(data, file)`: Writes a Python dictionary into a JSON file.
  * `json.load(file)`: Reads a JSON file and turns it back into a Python dictionary.


```python
import csv
import json

print("---- Working with CSV ----")
# Writing to a CSV

with open("users.csv","w", newline="") as file:
  writer = csv.writer(file)
  writer.writerow(["Username","Role","Status"])
  writer.writerow(["prashat_k","Admin","Active"])
  writer.writerow(["kumar_p","User","Disabled"])

# Reading from a CSV
with open("users.csv","r") as file:
  reader = csv.reader(file)
  for row in reader:
    print(row)

print("\n--- Working with JSON ----")

# A standard Python Dictionary (perfect for API payloads)
config_data = {
    "server_name": "EIC_Prod_01",
    "ip_address": "192.168.1.100",
    "timeout_seconds": 300
}

# Writing the dictionary to a JSON file
with open("config.json","w") as file:
  json.dump(config_data, file, indent=4)    # indent=4 makes it pretty and readable

# Reading the JSON file back into Python
with open("config.json","r") as file:
  loaded_data = json.load(file)

print(f"Loaded Server Name: {loaded_data['server_name']}")
print(f"Loaded Timeout: {loaded_data['timeout_seconds']}")
```

### Module 6.2: Modules, Packages & Virtual Environments (pip)
Using external libraries and managing project dependencies cleanly.


#### 1. Modules
A **module** is simply a `.py` file full of code you can reuse. Instead of rewriting everything, you **import** it. Just use `import module_name`.
* **Built-in modules** come with Python: `math`, `random`, `datetime`, `json`, `csv`.
* **Your own modules:** any `.py` file you write can be imported by another file.

| Style | Example | Use it as |
|---|---|---|
| Import whole module | `import math` | `math.sqrt(16)` |
| Import specific item | `from math import sqrt` | `sqrt(16)` |
| Import with nickname | `import random as rd` | `rd.randint(1, 6)` |


#### 2. Packages
A **package** is a **folder of modules**. Python treats a folder as a package when it contains a file named `__init__.py`. Think of a module as a book and a package as a bookshelf.

Third-party code. Download it first using `pip install package_name`, then `import` it.

#### 3. pip (The Package Installer)
**pip** downloads **external libraries** written by other people from PyPI (the Python Package Index).

* `pip install requests`: install a library
* `pip uninstall requests`: remove it
* `pip list`: see what's installed
* `pip freeze > requirements.txt`: save the exact list of libraries
* `pip install -r requirements.txt`: install everything from that list

#### 4. Virtual Environments

Project A may need version 1 of a library while Project B needs version 2. A **virtual environment (venv)** gives every project its **own private set of libraries**, so projects never clash.

### Module 6.3: Working with APIs & Automation (`requests`)
Fetching live data from the web, automating repetitive tasks, and parsing logs or user records.

The `requests` library allows Python to communicate with external web servers and REST APIs.



#### 1. What is an API?
An **API** (Application Programming Interface) is like a **waiter in a restaurant**: you (your program) give an order, the waiter takes it to the kitchen (a server), and brings back the result. Most web APIs reply with **JSON**, which you already know how to handle.




#### 2. The `requests` Library

`requests` is an external library (install with `pip install requests`) that talks to web APIs.

| Part | Meaning |
|---|---|
| `requests.get(url)` | Ask the server for data |
| `response.status_code` | `200` = success, `404` = not found, `500` = server error |
| `response.json()` | Convert the reply into a Python dict/list |
| `params={...}` | Add filters to the request (like `?userId=1`) |
| `timeout=10` | Give up after 10 seconds instead of waiting forever |
| `response.raise_for_status()` | Raise an error automatically if the request failed |

**Always** wrap network calls in `try / except`, because the internet can fail at any time.

#### 3. Automation

Automation means making Python do repetitive jobs for you. A typical pattern:

1. **Fetch** data (from an API or file)
2. **Process** it (loop, filter, count)
3. **Save** the result (CSV, JSON, or a report)

#### 4. Parsing Logs

A **log** is a text file where programs record what happened. Python can read every line, split it into parts, and count errors automatically, a job that would take a human hours.

**Tip:** In this module we use `jsonplaceholder.typicode.com`, a free practice API that needs no login.

```python
import requests
import csv

# ---------- 1. Fetch live data ----------
url = "https://jsonplaceholder.typicode.com/users"

try:
    response = requests.get(url, timeout=10)
    response.raise_for_status()          # error if status is not 2xx
    users = response.json()              # list of dictionaries
    print("Status:", response.status_code)       # 200
    print("Users received:", len(users))         # 10
    print("First user:", users[0]["name"])
except requests.exceptions.RequestException as e:
    print("Network problem:", e)
    users = []


# ---------- 2. Using params (filters) ----------
posts = requests.get(
    "https://jsonplaceholder.typicode.com/posts",
    params={"userId": 1},
    timeout=10,
).json()
print("Posts by user 1:", len(posts))


# ---------- 3. Automation: save the user records to a CSV ----------
if users:
    with open("users.csv", "w", newline="") as f:
        writer = csv.writer(f)
        writer.writerow(["id", "name", "email", "city"])
        for u in users:
            writer.writerow([u["id"], u["name"], u["email"], u["address"]["city"]])
    print("Saved users.csv")


# ---------- 4. Parsing a log file ----------
log_text = """2026-10-26 09:00:01 INFO Server started
2026-10-26 09:05:12 WARNING Disk space low
2026-10-26 09:07:45 ERROR Database connection failed
2026-10-26 09:08:02 INFO Retrying connection
2026-10-26 09:08:30 ERROR Database connection failed
"""

with open("app.log", "w") as f:
    f.write(log_text)

counts = {"INFO": 0, "WARNING": 0, "ERROR": 0}

with open("app.log", "r") as f:
    for line in f:
        parts = line.split()             # split into words
        level = parts[2]                 # the 3rd word is the level
        if level in counts:
            counts[level] += 1

print(counts)                            # {'INFO': 2, 'WARNING': 1, 'ERROR': 2}
```
