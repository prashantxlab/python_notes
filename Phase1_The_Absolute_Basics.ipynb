# Variables & Memory (How Python Remembers Things)

Right now, when we write print("CampusX"), Python displays it on the screen and immediately forgets it. If you need to use that name 50 times in your program, you would have to type it 50 times.
:To make Python remember data, we use Variables.

**1. What is a Variable?**
Think of your computer's memory (RAM) as a giant room full of empty storage boxes. A Variable is simply a named label you stick on a box to store a piece of data inside it.

```python
name = "Prashant"
age = 24
height = 5.9
is_learning = True
```

Look at that = sign. In school math, = means "equals to". In programming, = is called the Assignment Operator. It means: "Take whatever is on the right side, and store it inside the variable name on the left side."

Now, whenever you ask Python to print(name), Python looks up the box labeled name and shows what is inside:

```python
print(name)
print(age)
```

**NOTE:** we did not put quotes around name and age inside print(). If we put the quotes around the name and age it will literally print the name and age, When we write print(name) without quotes, Python fetches the value stored inside the variable name.

# Three Special Powers of Python Variables (CampusX Core Concepts)
**A. Dynamic Typing** (No Need to Declare Data Types)
In older languages like C++ or Java, before storing a number, you have to tell the computer: int age = 24.
Python is smart—it uses Dynamic Typing. You just write age = 24, and Python automatically figures out that 24 is an Integer (int)!

**B. Dynamic Binding** (Variables Can Change Their Type Anytime)
In Python, a variable is not locked to one data type. Look at this:

```python
x=10
print(x)

x="Hello"
print(x)
```

**NOTE:** Remember earlier we said Python reads code top to bottom, line by line? On line 1, x is 10. By line 4, x is overwritten with "Hello".

**C. Multiple Assignment Tricks**
You can create multiple variables in a single line:

```python
# Give different values to 3 variables at once:
a, b, c = 10, 20, 30
print(a,b,c)

# Give the SAME value to 3 variables at once:
x=y=z = 100
print(x,y,z)
```

# The 4 Strict Rules for Naming Variables

1.   Only Letters, Numbers, and Underscores (_): You cannot use spaces or special
symbols like @, $, -, or %.

user_name = "Rahul" (Valid — use underscore _ instead of space)

user-name = "Rahul" (Invalid — hyphen is not allowed)

# User Input (input) & Type Conversion
So far, we have been typing all the values directly inside the code (age = 24).
In real apps—like when you enter an OTP on Paytm or type your username on Instagram—the user types the data while the program is running.


1.   **Taking Input from the User: input()**
Whenever you want your Python program to pause, show a message, and wait for the user to type something on their keyboard, you use input().

```python
user_name = input("Enter your name: ")
print("Welcome",user_name,"!!!")
```

**What happens when this runs:**


*   Python displays Enter your name:  and waits with a blinking typing cursor.
*   Suppose you type Prashant and press Enter.
*   Python takes "Prashant", stores it inside the variable user_name, and moves to Line 2.
*   Line 2 prints: Welcome Prashant.

2. **The Giant Trap of input()**
Suppose we want to build a tiny calculator that asks the user for two numbers and adds them together. Look at what happens if we write this:

```python
num1 = input("Enter first number")
num2 = input("Enter second number")

result = num1+num2
print(result)
print("You would expect the output to be 30, right?")
print("Wrong! The output on the screen will be 1020!")
```

Why did 10 + 20 become 1020?!
Because of Python's Universal Rule of input():


*   No matter what the user types into input()—whether they type words, 10, 99.5, or True—Python ALWAYS receives it as a String (str)!

So when you typed 10 and 20, input() secretly turned them into "10" and "20" (Strings with quotes!).



3. **The Solution: Type Conversion (Type Casting):**
If input() always gives us a String ("10"), how do we do math with it? We convert it from a String into an Integer!
Converting one data type into another is called Type Conversion. Python gives you 4 simple commands named right after the data types:


*   int() -> Converts a value into an Integer (int("10") becomes 10)
*   float() -> Converts a value into a Float (float("10.5") becomes 10.5, and float(10) becomes 10.0)
*   str() -> Converts a value into a String (str(100) becomes "100")
*   bool() -> Converts a value into a Boolean





**Fixing Our Calculator with int():**
You can wrap int() directly around input() so the text gets converted into a number the exact millisecond the user presses Enter:

```python
num1 = int(input("Enter first number:"))
num2 = int(input("Enter second number:"))

result = num1 + num2
print(result)
```

**4. Implicit vs. Explicit Type Conversion**


1.   **Implicit Type Conversion (Automatic):** Python does this on its own without you asking, whenever it is safe and prevents data loss.



```python
a = 5   # int
b = 7.5   # float
c = a+b   # Python automatically converts 5 to 5.0 and gives 7.5 (float)

print(c)
```

2.   **Explicit Type Conversion (Manual / Type Casting):** When you force Python to convert a data type using int(), float(), or str().

*   Why doesn't Python automatically add "100" + 50? If you run print("100" + 50), Python crashes with a TypeError because it doesn't know whether you want math (150) or text ("10050"). You must explicitly tell it: int("100") + 50.
*   **NOTE:** You can only convert valid numbers! int("50") works (50), but int("Hello") will crash because "Hello" is not a number. Also, converting a float to an int (int(9.9)) chops off the decimal and gives 9.



# Operators in Python

Now that we know how to store data in variables and take user input, how do we actually do things with that data—like calculate a discount, check if a password matches, or verify if someone is eligible for an account?

We use Operators. Think of Operators as the action symbols in Python. We divides them into 4 main families that you will use every single day:


1.  **Arithmetic Operators (Math Calculations):** You already know + (addition), - (subtraction), and * (multiplication).
Wait—what about division? In Python, there are three different division operators and one power operator that every programmer must know by heart:

| Symbol | Name | What It Does | Example | Result |
| :--- | :--- | :--- | :--- | :--- |
| `+` | Addition | Adds two numbers | `10 + 3` | `13` |
| `-` | Subtraction | Subtracts two numbers | `10 - 3` | `7` |
| `*` | Multiplication | Multiplies two numbers | `10 * 3` | `30` |
| `/` | **Normal Division** | Divides two numbers. **Rule:** `/` *always* gives a **Float** (decimal), even if it divides evenly! | `10 / 2` <br> `7 / 2` | `5.0` <br> `3.5` |
| `//` | **Floor Division (Integer Division)** | Divides and **chops off the decimal part**, giving only the whole number part. | `7 // 2` <br> `10 // 3` | `3` <br> `3` |
| `%` | **Modulus (Remainder)** | Divides and gives you ONLY the **remainder** left over after division. | `7 % 2` <br> `10 % 2` | `1` <br> `0` |
| `**` | **Exponent (Power)** | Raises the first number to the power of the second number ($a^b$). | `2 ** 3` <br> `5 ** 2` | `8` ($2^3$) <br> `25` ($5^2$) |




2.   **Comparison (Relational) Operators (Asking Yes/No Questions):** Comparison operators compare two values and ALWAYS give a Boolean (True or False) as the answer.

| Symbol | Meaning | Example | Output |
| :--- | :--- | :--- | :--- |
| `==` | Equal to | `5 == 5` | `True` |
| `!=` | Not equal to | `5 != 10` | `True` |
| `>` | Greater than | `10 > 5` | `True` |
| `<` | Less than | `10 < 5` | `False` |
| `>=` | Greater than or equal to | `10 >= 10` | `True` |
| `<=` | Less than or equal to | `5 <= 3` | `False` |

The #1 Beginner Mistake (= vs ==):

**Single = (Assignment):** Stores a value in a box. x = 10 means "Put 10 inside x".

**Double == (Comparison):** Asks a question. x == 10 means "Hey Python, is the value inside x equal to 10? Tell me True or False."


**3.   Logical Operators (and, or, not):** What if you need to check two conditions at the same time? (For example: To log in, the username must be correct AND the password must be correct).

* **`and` (Strict Boss):** Gives `True` **ONLY if BOTH sides are `True`**. If even one side is `False`, the whole thing becomes `False`.
  * `(10 > 5) and (20 > 10)` -> `True and True` -> **`True`**
  * `(10 > 5) and (2 > 10)` -> `True and False` -> **`False`**
* **`or` (Easygoing Boss):** Gives `True` if **AT LEAST ONE side is `True`**. It only gives `False` when both sides are `False`.
  * `(10 > 5) or (2 > 10)` -> `True or False` -> **`True`**
  * `(1 > 5) or (2 > 10)` -> `False or False` -> **`False`**
* **`not` (The Inverter):** Flips `True` into `False`, and `False` into `True`.
  * `not(True)` -> **`False`**
  * `not(10 > 5)` -> `not(True)` -> **`False`**


**4.   Assignment Shortcuts (+=, -=) & Membership (in):**

**A. Updating a Variable (+=):** Suppose you have score = 10 and you want to add 5 to it.


*   **Long way:** score = score + 5 (Take old score 10, add 5, store 15 back in score).
*   **Pro Shortcut:** score += 5 (Means the exact same thing: add 5 directly to score!). You can also do score -= 5 or score *= 2.

**B. Membership Operator (in and not in):** Checks if a smaller piece exists inside a bigger item (like checking if the letter "a" is inside "Python"), and returns True or False:



```python
print("P" in "Python")       # Output: True
print("z" in "Python")       # Output: False
print("z" not in "Python")   # Output: True
```

```python
print(15 / 3)

print(17 // 5)

print(17 % 5)
```

* **Extracting vs Removing the Last Digit of a Number:**
  * `459 % 10` -> Gives `9` (**Extracts** the last digit using Remainder)
  * `459 // 10` -> Gives `45` (**Removes** the last digit using Floor Division)
