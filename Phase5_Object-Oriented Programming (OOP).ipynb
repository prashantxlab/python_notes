## Phase 5: Object-Oriented Programming

Object-Oriented Programming (OOP) allows you to move beyond built-in data types (like Lists and Dictionaries) and create your own custom data structures. Instead of just storing a user's name in a string and their age in an integer, OOP lets you create a custom User type that holds all their data and actions in one package.


### Module 5.1: Classes & Objects

* **Class:** The blueprint or template (e.g., the architectural drawing of a house).
* **Object:** The actual entity built from the blueprint (e.g., the physical house).
* **Methods:** The actions an object can do (a student can `introduce()` themselves). These are just functions written inside a class.
* **`__init__` (Constructor):** A special method that runs automatically when an object is created. It initializes the object's attributes.
* **Attributes:** Variables that belong to an object.
* **`self`:** Represents the specific object being created. It is how the object refers to its own attributes (translates to "my").

| Term | Real-life example |
|---|---|
| Class | The blueprint of a house |
| Object | One actual house built from it |
| Attribute | The house's color, number of rooms |
| Method | Opening the door, turning on the lights |

**Naming tip:** Class names use CapitalizedWords (`Student`), while variables and functions use lowercase (`student1`).


```python
# 1. Defining the Class (The Blueprint)
class User:
  # The Constructor: Runs automatically when we create a new user
  def __init__(self, username, role):
    self.username = username
    self.role = role
    print(f"--> A new User Object was created for {self.username}!")

print("Blueprint created. Now let's build some actual objects.\n")

# 2. Creating Objects (Building from the blueprint)
# We do not pass anything for 'self'—Python handles that automatically.

user1 = User("prashant_k","Admin")
user2 = User("kumar_p","Standard User")

print("\n--- Accessing Object Attributes ---")
# 3. Accessing the attributes using the dot (.)

print(f"User 1: {user1.username} | Role: {user1.role}")
print(f"User 2: {user2.username} | Role: {user2.role}")
```

```python
class Student:
  # __init__ runs automatically when an object is created
  def __init__(self, name, age):
    self.name = name        # attribute
    self.age = age          # attribute

  # a method: an action the object can perform
  def introduce(self):
    print(f"Hi, I am {self.name} and I am {self.age} years old")

  def birthday(self):
    self.age += 1
    print(f"Happy birthday {self.name}! Now {self.age}")


# Creating objects (instances) from the class
s1 = Student("Asha",20)
s2 = Student("Ravi",22)

s1.introduce()        # Hi, I'm Asha and I'm 20 years old.
s2.introduce()        # Hi, I'm Ravi and I'm 22 years old.

s1.birthday()         # Happy birthday Asha! Now 21.

# Each object keeps its OWN data
print(s1.age)      # 21
print(s2.age)      # 22  (unchanged)
```

#### Methods (Object Actions)
Attributes hold data (nouns), while **Methods** perform actions (verbs).
* A method is just a function defined inside a class.
* **The `self` Rule:** The first parameter of any method must be `self`. This gives the method permission to access the object's attributes.

```python
class User:
  def __init__(self, username):
    self.username = username
    self.is_logged_in = False     # Default status when a user is created

  # Method 1: Action to log in
  def login(self):
    self.is_logged_in = True
    print(f"Success: {self.username} has logged in.")

  # Method 2: Action to check status
  def check_status(self):
    if self.is_logged_in:
      print(f"{self.username} is currently ONLINE.")
    else:
      print(f"{self.username} is OFFLINE.")

# 1. Create the object
user1 = User("prashant_k")

# 2. Call methods using the dot (.)
user1.check_status()        # Starts offline
user1.login()               # Changes status
user1.check_status()        # Now online

```

### Module 5.2: Encapsulation (Protecting Data)

Protecting internal data and using getters/setters.

#### 1. Encapsulation (Protecting Your Data)
Right now, anyone can write `account.balance = -99999` and break your program. **Encapsulation** means hiding an object's internal data and only allowing access through controlled methods, like a bank where you cannot walk into the vault and must go through the cashier.

**Making an attribute private:** put two underscores `__` before its name.

* `self.balance`: **public**, anyone can read or change it.
* `self.__balance`: **private**, it cannot be accessed directly from outside the class.

#### 2. Getters and Setters

Since private data is locked, we create two kinds of methods to access it safely:

* **Getter:** A method that **reads** the value (`get_balance()`).
* **Setter:** A method that **changes** the value, and can **check it first** (reject negative money, for example).

#### 3. Reference Variables

When you write `b = a` for an object, Python does **not** make a copy. Both names point to the **same object** in memory, like two keys to the same locker.

* Changing the object through `a` is also visible through `b`.
* `a is b` checks whether both names point to the same object.
* To get a real, separate copy, use `.copy()` (for lists) or create a new object.

**Remember:** A variable is a **label** that points to an object. It is not the object itself.



```python
class BankAccount:
  def __init__(self, owner, starting_balance):
    self.owner = owner
    self.__balance = starting_balance     # The '__' makes this strictly PRIVATE

  # GETTER: Safely read the private data
  def get_balance(self):
    return self.__balance

  # SETTER: Safely modify the private data with validation rules
  def deposit(self, amount):
    if amount > 0:
      self.__balance += amount
      print(f"Success! Deposited ${amount}. New balance: ${self.__balance}")
    else:
      print("Security Alert: Deposit amount must be greater than zero.")

# 1. Create the account
my_account = BankAccount("Prashant", 5000)

# 2. Accessing a public attribute works fine
print("Account Owner:", my_account.owner)

# 3. Accessing a private attribute directly will CRASH the program with an AttributeError
# print(my_account.__balance)  #<-- Try uncommenting this to see the crash!

# 4. Instead, we use the safe Getter and Setter methods
my_account.deposit(2000)      # Valid data, works perfectly
my_account.deposit(-500)      # Invalid data, blocked by the setter
```

### Module 5.3: Inheritance & Polymorphism
Reusing code across classes and customizing behavior.

#### 1. Inheritance (Code Reuse):
Imagine you have a Vehicle class that knows how to start an engine and brake. If you want to build a Car class and a Motorcycle class, you don't rewrite that code. You just tell them to inherit those abilities from Vehicle, and then you only write the new stuff (like adding a sidecar to the motorcycle).

* The Child gets everything the Parent has automatically.
* **`super().__init__()`:** This built-in command calls the Parent's constructor to set up the inherited data, so you do not have to write it again.



#### 2. Polymorphism (Customizing Behavior)
This fancy word just means "many forms." It allows the child class to overwrite a parent's rule. If the Vehicle class has a `honk()` method that goes "Beep!", the Truck class can inherit it but override it to go "HOOONNNNK!".

*   **Polymorphism** allows the Child class to override or modify a method it inherited from the Parent to do something specific.



```python
# The Parent Class
class Employee:
  def __init__(self,name,salary):
    self.name = name
    self.salary = salary

  def work(self):
    print(f"{self.name} is doing regular employee tasks.")

# The Child Class (Inherits from Employee by passing it in the parentheses)
class Manager(Employee):
  def __init__(self, name, salary, department):
    # super() triggers the Parent's setup for name and salary
    super().__init__(name, salary)
    self.department = department      # A new attribute just for Managers

  # POLYMORPHISM: Overriding the Parent's work method
  def work(self):
    print(f"{self.name} is managing the {self.department} department.")

# 1. Create the objects
emp1 = Employee("Rahul",50000)
mgr1 = Manager("Anita",90000, "IT Security")

# 2. Call the methods
emp1.work()   # Runs the Parent version
mgr1.work()   # Runs the overridden Child version

# 3. Prove inheritance works
print(f"{mgr1.name} makes ${mgr1.salary} (Inherited Sucessfully)")
```
